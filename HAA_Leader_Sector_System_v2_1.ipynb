# 5. 공통 Utility

````python

def retry_call(func, *args, retries=None, sleep_sec=None, **kwargs):
    retries = retries or CONFIG["api_max_retries"]
    sleep_sec = sleep_sec or CONFIG["api_retry_sleep_sec"]

    last_error = None

    for attempt in range(1, retries + 1):
        try:
            return func(*args, **kwargs)
        except Exception as e:
            last_error = e
            if attempt < retries:
                time.sleep(sleep_sec * attempt)

    raise last_error


def cache_is_fresh(path, ttl_hours=None):
    ttl_hours = ttl_hours or CONFIG["cache_ttl_hours"]
    p = Path(path)

    if not p.exists():
        return False

    return (time.time() - p.stat().st_mtime) <= ttl_hours * 3600


def safe_write_csv(df, path, index=True):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    tmp = path.with_suffix(path.suffix + ".tmp")
    df.to_csv(tmp, index=index, encoding="utf-8-sig")
    tmp.replace(path)


def download_close(tickers, period="1y"):
    tickers = list(tickers)
    if not tickers:
        return pd.DataFrame()

    raw = yf.download(
        tickers,
        period=period,
        auto_adjust=True,
        progress=False,
        threads=True
    )

    if raw.empty:
        return pd.DataFrame()

    if isinstance(raw.columns, pd.MultiIndex):
        close = raw["Close"].copy()
    else:
        close = raw[["Close"]].copy()
        close.columns = tickers[:1]

    if isinstance(close, pd.Series):
        close = close.to_frame()

    return close.dropna(how="all")


def return_n_days(s, n):
    s = s.dropna()
    if len(s) <= n:
        return np.nan
    return s.iloc[-1] / s.iloc[-(n+1)] - 1


def momentum_13612u(s):
    vals = [return_n_days(s, n) for n in [21,63,126,252]]
    vals = [v for v in vals if pd.notna(v)]
    return np.mean(vals) if vals else np.nan


def find_col(df, keyword):
    for c in df.columns:
        if keyword in str(c):
            return c
    return None


def safe_pct_rank(series):
    s = pd.Series(series)
    return s.rank(pct=True) if s.notna().sum() else pd.Series(np.nan, index=s.index)


def classify_sector_from_name(name):
    name_u = str(name).upper()
    for sector, kws in SECTOR_KEYWORDS.items():
        for kw in kws:
            if str(kw).upper() in name_u:
                return sector
    return None


def is_excluded_etf(name):
    name_u = str(name).upper()
    return any(str(k).upper() in name_u for k in EXCLUDE_KEYWORDS)


class KRXAuthenticationError(RuntimeError):
    pass


class KRXDataError(RuntimeError):
    pass


class MarketDataError(RuntimeError):
    pass


_KRX_PREFLIGHT_RESULT = None


def normalize_krx_ticker(ticker):
    value = str(ticker).strip().upper()
    if not value or len(value) != 6 or not value.isalnum():
        raise KRXDataError(f"Invalid KRX ticker format: {ticker!r}")
    return value


def require_krx_credentials():
    id_env = CONFIG["krx_id_env"]
    pw_env = CONFIG["krx_pw_env"]
    missing = [
        name for name in [id_env, pw_env]
        if not os.environ.get(name, "").strip()
    ]
    if missing:
        raise KRXAuthenticationError(
            "KRX credentials are missing. Set environment variables: "
            + ", ".join(missing)
        )
    return {"id_env": id_env, "pw_env": pw_env}


def krx_preflight(force=False):
    global _KRX_PREFLIGHT_RESULT

    if _KRX_PREFLIGHT_RESULT is not None and not force:
        return _KRX_PREFLIGHT_RESULT.copy()

    require_krx_credentials()

    test_date = CONFIG["krx_preflight_date"]
    expected_count = int(CONFIG["krx_expected_etf_count"])
    mixed_probe = normalize_krx_ticker(CONFIG["krx_mixed_ticker_probe"])

    try:
        raw = retry_call(stock.get_etf_ticker_list, test_date)
    except Exception as e:
        raise KRXDataError(
            f"KRX ETF preflight request failed for {test_date}: {type(e).__name__}: {e}"
        ) from e

    tickers = [normalize_krx_ticker(t) for t in raw]

    if len(tickers) != expected_count:
        raise KRXDataError(
            f"KRX ETF preflight count mismatch for {test_date}: "
            f"expected {expected_count:,}, received {len(tickers):,}. "
            "No fallback data will be used."
        )

    if mixed_probe not in tickers:
        raise KRXDataError(
            f"KRX mixed ticker probe {mixed_probe} is missing from the "
            f"{test_date} ETF universe."
        )

    _KRX_PREFLIGHT_RESULT = {
        "status": "ok",
        "date": test_date,
        "etf_count": len(tickers),
        "expected_etf_count": expected_count,
        "mixed_ticker_probe": mixed_probe,
        "credentials_present": True,
    }
    return _KRX_PREFLIGHT_RESULT.copy()


def is_krx_etf_trading_date(date, fetch_snapshot=None):
    """Return True only for a weekday with a valid KRX ETF OHLCV snapshot."""
    parsed = datetime.strptime(str(date), "%Y%m%d")
    if parsed.weekday() >= 5:
        return False

    if fetch_snapshot is None:
        fetch_snapshot = lambda ds: retry_call(
            stock.get_etf_ohlcv_by_ticker,
            ds,
            retries=CONFIG["api_max_retries"],
        )

    snap = fetch_snapshot(str(date))
    required = {"종가", "거래대금"}
    if snap is None or snap.empty or not required.issubset(snap.columns):
        return False

    close = pd.to_numeric(snap["종가"], errors="coerce").fillna(0)
    trading_value = pd.to_numeric(snap["거래대금"], errors="coerce").fillna(0)
    return bool(close.gt(0).any() and trading_value.gt(0).any())


def nearest_business_date_str(days_back=10):
    errors = []

    for i in range(days_back + 1):
        date = (datetime.now() - timedelta(days=i)).strftime("%Y%m%d")
        try:
            if is_krx_etf_trading_date(date):
                return date
            errors.append(f"{date}: not a valid KRX ETF trading session")
        except Exception as e:
            errors.append(f"{date}: {type(e).__name__}: {e}")

    detail = " | ".join(errors[-3:])
    raise KRXDataError(
        "Unable to resolve a live KRX ETF business date from validated OHLCV. "
        f"No fallback date was used. Recent errors: {detail}"
    )


def download_adjusted_close(tickers, period="2y"):
    tickers = [str(t).strip() for t in tickers]
    if not tickers:
        raise MarketDataError("Adjusted Close request requires at least one ticker.")

    raw = yf.download(
        tickers,
        period=period,
        auto_adjust=False,
        actions=False,
        progress=False,
        threads=True,
    )

    if raw is None or raw.empty:
        raise MarketDataError(
            "Yahoo Finance returned no data for Adjusted Close / Total Return."
        )

    if isinstance(raw.columns, pd.MultiIndex):
        level0 = set(raw.columns.get_level_values(0))
        if "Adj Close" not in level0:
            raise MarketDataError(
                "Yahoo Finance response has no 'Adj Close' field; "
                "unadjusted Close will not be substituted."
            )
        close = raw["Adj Close"].copy()
    else:
        if "Adj Close" not in raw.columns:
            raise MarketDataError(
                "Yahoo Finance response has no 'Adj Close' field; "
                "unadjusted Close will not be substituted."
            )
        close = raw[["Adj Close"]].copy()
        close.columns = tickers[:1]

    if isinstance(close, pd.Series):
        close = close.to_frame(name=tickers[0])

    missing = [
        ticker for ticker in tickers
        if ticker not in close.columns or close[ticker].dropna().empty
    ]
    if missing:
        raise MarketDataError(
            "Adjusted Close is missing for: " + ", ".join(missing)
        )

    return close.dropna(how="all")


def classify_haa_risk_state(tip_momentum, tip_1m, tip_3m):
    risk_on = bool(pd.notna(tip_momentum) and tip_momentum > 0)
    weak_risk_on = bool(
        risk_on
        and pd.notna(tip_1m) and tip_1m < 0
        and pd.notna(tip_3m) and tip_3m < 0
    )

    if weak_risk_on:
        state = "WEAK_RISK_ON"
    elif risk_on:
        state = "RISK_ON"
    else:
        state = "RISK_OFF"

    return {
        "risk_on": risk_on,
        "weak_risk_on": weak_risk_on,
        "risk_state": state,
    }


if os.environ.get("HAA_AUTO_RUN_KRX_PREFLIGHT", "0") == "1":
    AUTO_KRX_PREFLIGHT_RESULT = krx_preflight(force=True)
    print(
        "AUTO_KRX_PREFLIGHT_OK "
        f"date={AUTO_KRX_PREFLIGHT_RESULT['date']} "
        f"etf_count={AUTO_KRX_PREFLIGHT_RESULT['etf_count']} "
        f"mixed_ticker={AUTO_KRX_PREFLIGHT_RESULT['mixed_ticker_probe']}"
    )
````

# 6. Risk Layer — HAA + Fear & Greed + KOSPI

````python

CNN_FG_URL = "https://production.dataviz.cnn.io/index/fearandgreed/graphdata"

def get_haa_state():
    tickers = sorted(set(
        [CONFIG["haa_canary"]] +
        CONFIG["haa_offensive"] +
        CONFIG["haa_defensive"]
    ))

    close = download_adjusted_close(tickers, period="2y")

    missing = [t for t in tickers if t not in close.columns]
    if missing:
        raise MarketDataError(
            "HAA Adjusted Close is missing for: " + ", ".join(missing)
        )

    mom = pd.Series({
        t: momentum_13612u(close[t].dropna())
        for t in tickers
    }).sort_values(ascending=False)

    canary = CONFIG["haa_canary"]
    tip_close = close[canary].dropna()
    tip_momentum = mom.get(canary, np.nan)
    tip_1m = return_n_days(tip_close, 21)
    tip_3m = return_n_days(tip_close, 63)
    state = classify_haa_risk_state(tip_momentum, tip_1m, tip_3m)

    return {
        "close": close,
        "price_basis": "Yahoo Finance Adj Close (split/dividend adjusted total-return series)",
        "momentum": mom,
        "tip_momentum": tip_momentum,
        "tip_1m": tip_1m,
        "tip_3m": tip_3m,
        **state,
    }


def get_cnn_fear_greed():
    if CONFIG["use_cnn_fear_greed"]:
        try:
            r = requests.get(
                CNN_FG_URL,
                headers={"User-Agent":"Mozilla/5.0"},
                timeout=10
            )
            r.raise_for_status()
            data = r.json()

            fg = data.get("fear_and_greed", {})

            for key in ["score","value","rating"]:
                v = fg.get(key)
                if isinstance(v, (int,float)) and 0 <= float(v) <= 100:
                    return {"value":float(v), "source":"CNN", "status":"ok"}
        except Exception as e:
            error = str(e)
        else:
            error = None
    else:
        error = None

    manual = CONFIG["manual_fear_greed"]

    if manual is not None:
        return {
            "value": float(manual),
            "source": "MANUAL",
            "status": "fallback"
        }

    return {
        "value": None,
        "source": None,
        "status": "missing",
        "error": error
    }


def get_kospi_close(period="1y"):
    df = yf.download(
        "^KS11",
        period=period,
        auto_adjust=True,
        progress=False
    )

    if isinstance(df.columns, pd.MultiIndex):
        s = df["Close"]
        if isinstance(s, pd.DataFrame):
            s = s.iloc[:,0]
    else:
        s = df["Close"]

    return s.dropna()


def tactical_cash_weight(risk_on, kospi, fear_greed=None):
    ma11 = kospi.rolling(11).mean()
    ma21 = kospi.rolling(21).mean()
    ret5 = kospi.pct_change(5)

    above11 = bool(kospi.iloc[-1] > ma11.iloc[-1])
    above21 = bool(kospi.iloc[-1] > ma21.iloc[-1])
    ret5_pos = bool(ret5.iloc[-1] > 0)

    reasons = []

    if not risk_on:
        cash = 0.30
        reasons.append("HAA Risk-Off")
    else:
        cash = 0.0
        reasons.append("HAA Risk-On")

        if not above11:
            cash = max(cash, 0.10)
            reasons.append("KOSPI < MA11")

        if not above21:
            cash = max(cash, 0.20)
            reasons.append("KOSPI < MA21")

        if (not above11) and (not above21) and (not ret5_pos):
            cash = 0.30
            reasons.append("KOSPI MA11/21 break + 5D weakness")

        if fear_greed is not None and pd.notna(fear_greed):
            fg = float(fear_greed)

            if fg >= CONFIG["fg_greed_level"]:
                cash = max(cash, 0.10)
                reasons.append("Fear&Greed greed")

            if fg >= CONFIG["fg_extreme_level"]:
                cash = max(cash, 0.20)
                reasons.append("Fear&Greed extreme")

            if (
                fg >= CONFIG["fg_max_risk_level"]
                and ((not above11) or (not ret5_pos))
            ):
                cash = max(cash, 0.30)
                reasons.append("Extreme greed + trend deterioration")

    return {
        "cash_weight": cash,
        "above11": above11,
        "above21": above21,
        "ret5_positive": ret5_pos,
        "fear_greed": fear_greed,
        "reasons": reasons,
    }
````

# 7. ETF Universe / Leader Sectors

````python

def get_all_kr_etfs(date=None):
    preflight = krx_preflight()

    if date is None:
        date = nearest_business_date_str()

    try:
        raw_tickers = retry_call(stock.get_etf_ticker_list, date)
    except Exception as e:
        raise KRXDataError(
            f"KRX ETF universe request failed for {date}: "
            f"{type(e).__name__}: {e}. No fallback data will be used."
        ) from e

    if not raw_tickers:
        raise KRXDataError(
            f"KRX ETF universe is empty for {date}. No fallback data will be used."
        )

    tickers = [normalize_krx_ticker(t) for t in raw_tickers]
    if len(set(tickers)) != len(tickers):
        raise KRXDataError(f"KRX ETF universe contains duplicate tickers for {date}.")

    rows = []
    name_errors = []

    for ticker in tickers:
        try:
            name = retry_call(stock.get_etf_ticker_name, ticker)
        except Exception as e:
            name_errors.append(f"{ticker}: {type(e).__name__}: {e}")
            continue

        if not str(name).strip():
            name_errors.append(f"{ticker}: empty ETF name")
            continue

        name = str(name).strip()
        sector = classify_sector_from_name(name)
        excluded = is_excluded_etf(name)

        rows.append({
            "ticker": ticker,
            "name": name,
            "sector": sector,
            "excluded": excluded,
            "is_sector_candidate": bool((sector is not None) and (not excluded)),
            "universe_date": date,
        })

    if name_errors:
        sample = " | ".join(name_errors[:5])
        raise KRXDataError(
            f"KRX returned incomplete ETF metadata for {len(name_errors)} of "
            f"{len(tickers)} tickers. No partial universe will be used. {sample}"
        )

    universe = pd.DataFrame(rows)

    if len(universe) != len(tickers):
        raise KRXDataError(
            f"KRX ETF universe normalization mismatch: "
            f"{len(tickers)} tickers, {len(universe)} rows."
        )

    universe.attrs["krx_preflight"] = preflight
    universe.attrs["full_universe_count"] = len(universe)
    universe.attrs["sector_candidate_count"] = int(
        universe["is_sector_candidate"].sum()
    )
    return universe


def build_kr_etf_rank():
    universe = get_all_kr_etfs()

    # 먼저 전체 ETF Universe를 검증한 뒤 섹터 ETF 후보군만 필터링한다.
    universe = universe[universe["is_sector_candidate"]].copy()

    if universe.empty:
        raise KRXDataError(
            "The validated KRX ETF universe contains no sector ETF candidates."
        )

    date = nearest_business_date_str()

    try:
        snap = retry_call(
            stock.get_etf_ohlcv_by_ticker,
            date,
            retries=CONFIG["api_max_retries"],
        )
    except Exception as e:
        raise KRXDataError(
            f"KRX ETF OHLCV snapshot request failed for {date}: "
            f"{type(e).__name__}: {e}. No unfiltered universe will be used."
        ) from e

    required_snapshot_columns = {"종가", "거래대금"}
    if (
        snap is None
        or snap.empty
        or not required_snapshot_columns.issubset(snap.columns)
    ):
        raise KRXDataError(
            f"KRX ETF OHLCV snapshot is invalid for {date}; "
            "required columns are 종가 and 거래대금. "
            "No unfiltered universe will be used."
        )

    snap = snap.copy()
    snap.index = snap.index.astype(str)
    universe = universe.merge(
        snap[["종가", "거래대금"]],
        left_on="ticker",
        right_index=True,
        how="left",
    )
    universe = universe[
        universe["거래대금"].fillna(0) >= CONFIG["min_etf_trading_value"]
    ]

    yahoo_map = {f"{t}.KS":str(t) for t in universe["ticker"]}
    close = download_close(yahoo_map.keys(), period="1y").rename(columns=yahoo_map)

    meta = universe.set_index("ticker")[["name","sector","거래대금"]].to_dict("index")
    rows = []

    for ticker in close.columns:
        ticker = str(ticker)

        if ticker not in meta:
            continue

        s = close[ticker].dropna()

        if len(s) < 70:
            continue

        r = s.pct_change().dropna()
        neg = r.tail(63)
        neg = neg[neg < 0]

        downside = neg.std(ddof=1) * np.sqrt(252) if len(neg) > 1 else np.nan
        ret63 = return_n_days(s, 63)

        rows.append({
            "ticker": ticker,
            "name": meta[ticker]["name"],
            "sector": meta[ticker]["sector"],
            "거래대금": meta[ticker]["거래대금"],
            "ret21": return_n_days(s,21),
            "ret63": ret63,
            "ret126": return_n_days(s,126),
            "sortino63": ret63/downside if pd.notna(downside) and downside > 0 else np.nan,
            "above_ma11": bool(s.iloc[-1] > s.rolling(11).mean().iloc[-1]),
            "above_ma21": bool(s.iloc[-1] > s.rolling(21).mean().iloc[-1]),
        })

    rank = pd.DataFrame(rows)

    if rank.empty:
        return rank, close

    for src, dst in [
        ("ret21","rs21_pct"),
        ("ret63","rs63_pct"),
        ("ret126","rs126_pct"),
        ("sortino63","sortino_pct"),
    ]:
        rank[dst] = rank[src].rank(pct=True)

    rank["momentum_score"] = (
        0.20*rank["rs21_pct"] +
        0.50*rank["rs63_pct"] +
        0.15*rank["rs126_pct"] +
        0.15*rank["sortino_pct"]
    )

    rank["trend_score"] = (
        0.6*rank["above_ma11"].astype(int) +
        0.4*rank["above_ma21"].astype(int)
    )

    rank["kr_score"] = (
        0.80*rank["momentum_score"] +
        0.20*rank["trend_score"]
    )

    return rank.sort_values("kr_score", ascending=False).reset_index(drop=True), close


def build_us_sector_rank():
    tickers = sorted({t for x in US_SECTOR_ETFS.values() for t in x})
    close = download_close(tickers, period="1y")
    rows = []

    for sector, names in US_SECTOR_ETFS.items():
        tmp = []

        for t in names:
            if t not in close.columns:
                continue

            s = close[t].dropna()

            if len(s) < 70:
                continue

            ret63 = return_n_days(s,63)
            r = s.pct_change().dropna().tail(63)
            neg = r[r < 0]
            dd = neg.std(ddof=1)*np.sqrt(252) if len(neg)>1 else np.nan

            tmp.append({
                "ret63": ret63,
                "sortino": ret63/dd if pd.notna(dd) and dd>0 else np.nan,
                "above11": s.iloc[-1] > s.rolling(11).mean().iloc[-1],
                "above21": s.iloc[-1] > s.rolling(21).mean().iloc[-1],
            })

        if not tmp:
            continue

        x = pd.DataFrame(tmp)

        rows.append({
            "sector": sector,
            "us_ret63": x["ret63"].mean(),
            "us_sortino": x["sortino"].mean(),
            "us_above11": bool(x["above11"].mean() >= 0.5),
            "us_above21": bool(x["above21"].mean() >= 0.5),
        })

    rank = pd.DataFrame(rows)

    if rank.empty:
        return rank

    rank["us_rs63_pct"] = rank["us_ret63"].rank(pct=True)
    rank["us_sortino_pct"] = rank["us_sortino"].rank(pct=True)

    rank["us_score"] = (
        0.60*rank["us_rs63_pct"] +
        0.20*rank["us_sortino_pct"] +
        0.12*rank["us_above11"].astype(int) +
        0.08*rank["us_above21"].astype(int)
    )

    return rank.sort_values("us_score", ascending=False).reset_index(drop=True)


def build_leaders_today():
    kr_rank, kr_close = build_kr_etf_rank()
    us_rank = build_us_sector_rank()

    if kr_rank.empty:
        return pd.DataFrame(), kr_rank, us_rank, kr_close

    reps = (
        kr_rank.sort_values("kr_score", ascending=False)
        .groupby("sector", as_index=False)
        .first()
        [["sector","ticker","name","kr_score","ret63","above_ma11","above_ma21"]]
        .rename(columns={
            "ticker":"rep_ticker",
            "name":"rep_name",
            "ret63":"kr_ret63"
        })
    )

    avgs = (
        kr_rank.groupby("sector", as_index=False)
        .agg(
            kr_sector_avg_score=("kr_score","mean"),
            kr_etf_count=("ticker","count")
        )
    )

    out = reps.merge(avgs, on="sector", how="left")

    if not us_rank.empty:
        out = out.merge(
            us_rank[["sector","us_score","us_ret63","us_above11","us_above21"]],
            on="sector",
            how="left"
        )

    out["us_score"] = out.get("us_score",0)
    out["us_score"] = out["us_score"].fillna(0)

    out["us_above11"] = out.get("us_above11",False)
    out["us_above11"] = out["us_above11"].fillna(False)

    out["cross_market_confirm"] = (
        out["us_above11"] &
        (out["us_score"] >= 0.60)
    )

    out["final_sector_score"] = (
        0.70*out["kr_score"] +
        0.15*out["kr_sector_avg_score"] +
        0.15*out["cross_market_confirm"].astype(int)
    )

    leaders = (
        out[out["above_ma11"]]
        .sort_values("final_sector_score", ascending=False)
        .head(CONFIG["leader_sector_count"])
        .reset_index(drop=True)
    )

    return leaders, kr_rank, us_rank, kr_close



def run_market_smoke_test():
    """실제 주문·Alpha·NAV 없이 시장 데이터와 주도섹터 경로만 검증한다."""
    haa = get_haa_state()
    kospi = get_kospi_close()
    leaders, kr_rank, us_rank, _ = build_leaders_today()

    if haa.get("price_basis") != (
        "Yahoo Finance Adj Close (split/dividend adjusted total-return series)"
    ):
        raise MarketDataError(
            "TIP/HAA price basis is not the required Adjusted Close series."
        )

    for key in ["tip_momentum", "tip_1m", "tip_3m"]:
        if pd.isna(haa.get(key)):
            raise MarketDataError(f"HAA smoke test produced missing {key}.")

    if haa.get("risk_state") not in {
        "RISK_ON", "WEAK_RISK_ON", "RISK_OFF"
    }:
        raise MarketDataError(
            f"Unexpected HAA risk state: {haa.get('risk_state')!r}"
        )

    if kospi is None or len(kospi.dropna()) < 21:
        raise MarketDataError("KOSPI smoke test requires at least 21 closes.")

    kr_required = {
        "ret21", "ret63", "ret126", "sortino63",
        "above_ma11", "above_ma21", "kr_score",
    }
    if kr_rank.empty or not kr_required.issubset(kr_rank.columns):
        missing = sorted(kr_required - set(kr_rank.columns))
        raise MarketDataError(
            f"KR ETF ranking smoke test failed; missing columns: {missing}"
        )

    us_required = {
        "us_ret63", "us_sortino", "us_above11",
        "us_above21", "us_score",
    }
    if us_rank.empty or not us_required.issubset(us_rank.columns):
        missing = sorted(us_required - set(us_rank.columns))
        raise MarketDataError(
            f"US sector cross-confirmation failed; missing columns: {missing}"
        )

    leader_required = {
        "sector", "rep_ticker", "rep_name",
        "cross_market_confirm", "final_sector_score",
    }
    if leaders.empty or not leader_required.issubset(leaders.columns):
        missing = sorted(leader_required - set(leaders.columns))
        raise MarketDataError(
            f"Leader-sector smoke test failed; missing columns: {missing}"
        )

    if len(leaders) > CONFIG["leader_sector_count"]:
        raise MarketDataError(
            f"Leader count {len(leaders)} exceeds configured Top "
            f"{CONFIG['leader_sector_count']}."
        )

    sector_names = ",".join(leaders["sector"].astype(str).tolist())
    print(
        "MARKET_SMOKE_TEST_OK "
        f"price_basis=ADJ_CLOSE "
        f"tip_13612u={haa['tip_momentum']:.6f} "
        f"tip_1m={haa['tip_1m']:.6f} "
        f"tip_3m={haa['tip_3m']:.6f} "
        f"risk_state={haa['risk_state']} "
        f"kr_etfs={len(kr_rank)} "
        f"us_sectors={len(us_rank)} "
        f"leaders={len(leaders)} "
        f"sectors={sector_names}"
    )

    return {
        "status": "ok",
        "haa": {
            "price_basis": haa["price_basis"],
            "tip_13612u": haa["tip_momentum"],
            "tip_1m": haa["tip_1m"],
            "tip_3m": haa["tip_3m"],
            "risk_state": haa["risk_state"],
        },
        "kr_rank_count": len(kr_rank),
        "us_rank_count": len(us_rank),
        "leaders": leaders[
            [
                "sector", "rep_ticker", "rep_name",
                "cross_market_confirm", "final_sector_score",
            ]
        ].copy(),
        "orders_created": False,
        "_haa": haa,
        "_kospi": kospi,
        "_leaders": leaders,
        "_kr_rank": kr_rank,
        "_us_rank": us_rank,
    }


if os.environ.get("HAA_EXECUTION_CHECK_ONLY", "0") == "1":
    MARKET_SMOKE_TEST_RESULT = run_market_smoke_test()
````

# 8. ETF Snapshot / Alpha

````python

ETF_CONSTITUENT_DIR = BASE_DIR / "etf_constituents"

def get_etf_constituents(etf_ticker, date=None):
    ticker = normalize_krx_ticker(etf_ticker)
    if date is None:
        date = nearest_business_date_str()

    try:
        # Public pykrx.stock API order is (ticker, date).
        df = retry_call(
            stock.get_etf_portfolio_deposit_file,
            ticker,
            str(date),
            retries=CONFIG["api_max_retries"],
        )
    except Exception as e:
        raise KRXDataError(
            f"KRX ETF constituent request failed for ticker={ticker}, "
            f"date={date}: {type(e).__name__}: {e}. "
            "No empty or fallback constituent set will be used."
        ) from e

    if df is None or len(df) == 0:
        raise KRXDataError(
            f"KRX ETF constituent response is empty for ticker={ticker}, "
            f"date={date}. No fallback constituent set will be used."
        )

    out = df.copy()
    out.index = out.index.astype(str)
    out = out[out.index.str.fullmatch(r"\d{6}", na=False)].copy()
    out["ticker"] = out.index

    if "구성종목명" not in out.columns:
        out["구성종목명"] = out["ticker"].apply(stock.get_market_ticker_name)

    return out.reset_index(drop=True)


def save_etf_snapshot(sector, etf_ticker, etf_name=None):
    date = nearest_business_date_str()
    df = get_etf_constituents(etf_ticker, date)

    if df.empty:
        return None

    date_tag = pd.Timestamp(date).strftime("%Y-%m-%d")

    df["sector"] = sector
    df["source_etf"] = str(etf_ticker)
    df["source_etf_name"] = etf_name
    df["snapshot_date"] = date_tag

    folder = ETF_CONSTITUENT_DIR / sector / str(etf_ticker)
    folder.mkdir(parents=True, exist_ok=True)

    path = folder / f"{date_tag}.csv"
    df.to_csv(path, index=False, encoding="utf-8-sig")

    return path


_KRX_LATEST_CLOSE_CACHE = None


def parse_krx_latest_close_snapshot(stock_raw, etf_raw, date):
    stock_required = {"ISU_SRT_CD", "TDD_CLSPRC"}
    if (
        stock_raw is None
        or stock_raw.empty
        or not stock_required.issubset(stock_raw.columns)
    ):
        raise KRXDataError(
            f"KRX listed-stock close snapshot is invalid for {date}."
        )

    if (
        etf_raw is None
        or etf_raw.empty
        or "종가" not in etf_raw.columns
    ):
        raise KRXDataError(
            f"KRX ETF close snapshot is invalid for {date}."
        )

    stocks = stock_raw[list(stock_required)].copy()
    stocks["ticker"] = (
        stocks["ISU_SRT_CD"].astype(str).str.strip().str.zfill(6)
    )
    stocks["price"] = pd.to_numeric(
        stocks["TDD_CLSPRC"].astype(str).str.replace(",", "", regex=False),
        errors="coerce",
    )
    stocks["source"] = "STOCK"

    etfs = etf_raw[["종가"]].copy()
    etfs.index = etfs.index.astype(str).str.strip().str.upper()
    etfs = etfs.reset_index()
    etfs.columns = ["ticker", "price"]
    etfs["price"] = pd.to_numeric(etfs["price"], errors="coerce")
    etfs["source"] = "ETF"

    out = pd.concat(
        [
            stocks[["ticker", "price", "source"]],
            etfs[["ticker", "price", "source"]],
        ],
        ignore_index=True,
    )
    out["ticker"] = out["ticker"].astype(str).str.strip().str.upper()
    out = out[
        out["ticker"].str.fullmatch(r"[0-9A-Z]{6}", na=False)
        & out["price"].gt(0)
    ].copy()

    duplicates = out.loc[
        out["ticker"].duplicated(keep=False), "ticker"
    ].unique()
    if len(duplicates):
        raise KRXDataError(
            "KRX stock/ETF close snapshot contains duplicate tickers: "
            + ", ".join(map(str, duplicates[:10]))
        )

    if out.empty:
        raise KRXDataError(
            f"KRX stock/ETF close snapshot has no valid prices for {date}."
        )

    return out.set_index("ticker").sort_index()


def get_krx_latest_close_snapshot(force=False):
    global _KRX_LATEST_CLOSE_CACHE

    date = nearest_business_date_str()
    if (
        _KRX_LATEST_CLOSE_CACHE is not None
        and not force
        and _KRX_LATEST_CLOSE_CACHE["date"] == date
    ):
        return _KRX_LATEST_CLOSE_CACHE["prices"].copy()

    try:
        stock_raw = retry_call(
            전종목시세().fetch,
            date,
            "ALL",
            retries=CONFIG["api_max_retries"],
        )
        etf_raw = retry_call(
            stock.get_etf_ohlcv_by_ticker,
            date,
            retries=CONFIG["api_max_retries"],
        )
    except Exception as e:
        raise KRXDataError(
            f"KRX stock/ETF close snapshot request failed for {date}: "
            f"{type(e).__name__}: {e}."
        ) from e

    prices = parse_krx_latest_close_snapshot(stock_raw, etf_raw, date)
    _KRX_LATEST_CLOSE_CACHE = {
        "date": date,
        "prices": prices,
    }
    return prices.copy()


def get_latest_close_krx(ticker):
    ticker = str(ticker).strip().upper()
    prices = get_krx_latest_close_snapshot()

    if ticker not in prices.index:
        raise KRXDataError(
            f"KRX latest close is missing for ticker={ticker}, "
            f"date={nearest_business_date_str()}."
        )

    price = float(prices.loc[ticker, "price"])
    if not np.isfinite(price) or price <= 0:
        raise KRXDataError(
            f"KRX latest close is invalid for ticker={ticker}: {price!r}."
        )
    return price

from pykrx.website.krx.market.core import (
    전종목기본정보,
    전종목시세,
    개별종목시세,
    투자자별_거래실적_개별종목_일별추이_일반,
)


_KRX_STOCK_MASTER = None


def get_krx_stock_master(force=False):
    """Load a validated short-ticker -> ISIN/name master from authenticated KRX."""
    global _KRX_STOCK_MASTER

    if _KRX_STOCK_MASTER is not None and not force:
        return _KRX_STOCK_MASTER.copy()

    try:
        raw = retry_call(
            전종목기본정보().fetch,
            "ALL",
            "ALL",
            retries=CONFIG["api_max_retries"],
        )
    except Exception as e:
        raise KRXDataError(
            "KRX listed-stock master request failed: "
            f"{type(e).__name__}: {e}. No inferred ISIN will be used."
        ) from e

    required = {"ISU_SRT_CD", "ISU_CD", "ISU_ABBRV"}
    if raw is None or raw.empty or not required.issubset(raw.columns):
        raise KRXDataError(
            "KRX listed-stock master is empty or missing ticker/ISIN/name. "
            "No inferred ISIN will be used."
        )

    master = raw[list(required)].copy()
    master["ticker"] = (
        master["ISU_SRT_CD"].astype(str).str.strip().str.zfill(6)
    )
    master["isin"] = master["ISU_CD"].astype(str).str.strip()
    master["name"] = master["ISU_ABBRV"].astype(str).str.strip()
    master = master[
        master["ticker"].str.fullmatch(r"\d{6}", na=False)
        & master["isin"].str.fullmatch(r"KR[A-Z0-9]{10}", na=False)
        & master["name"].ne("")
    ][["ticker", "isin", "name"]].copy()

    if master.empty or master["ticker"].duplicated().any():
        raise KRXDataError(
            "KRX listed-stock master normalization failed or contains "
            "duplicate tickers."
        )

    _KRX_STOCK_MASTER = master.set_index("ticker").sort_index()
    return _KRX_STOCK_MASTER.copy()


def _krx_numeric(series):
    return pd.to_numeric(
        series.astype(str).str.replace(",", "", regex=False),
        errors="coerce",
    )


def parse_krx_stock_ohlcv(raw, ticker, start_date, end_date):
    source_columns = {
        "TRD_DD": "날짜",
        "TDD_OPNPRC": "시가",
        "TDD_HGPRC": "고가",
        "TDD_LWPRC": "저가",
        "TDD_CLSPRC": "종가",
        "ACC_TRDVOL": "거래량",
        "ACC_TRDVAL": "거래대금",
        "MKTCAP": "시가총액",
    }
    if raw is None or raw.empty or not set(source_columns).issubset(raw.columns):
        raise KRXDataError(
            f"KRX stock OHLCV is empty or incomplete for ticker={ticker}, "
            f"range={start_date}:{end_date}."
        )

    out = raw[list(source_columns)].rename(columns=source_columns).copy()
    out["날짜"] = pd.to_datetime(out["날짜"], errors="coerce")
    for col in source_columns.values():
        if col != "날짜":
            out[col] = _krx_numeric(out[col])

    out = out.dropna(subset=["날짜", "종가", "거래대금", "시가총액"])
    out = out.set_index("날짜").sort_index()
    if out.empty or out["종가"].le(0).all():
        raise KRXDataError(
            f"KRX stock OHLCV has no valid closes for ticker={ticker}, "
            f"range={start_date}:{end_date}."
        )
    return out


def parse_krx_stock_flow(raw, ticker, start_date, end_date):
    source_columns = {
        "TRD_DD": "날짜",
        "TRDVAL1": "기관합계",
        "TRDVAL2": "기타법인",
        "TRDVAL3": "개인",
        "TRDVAL4": "외국인합계",
        "TRDVAL_TOT": "전체",
    }
    if raw is None or raw.empty or not set(source_columns).issubset(raw.columns):
        raise KRXDataError(
            f"KRX investor flow is empty or incomplete for ticker={ticker}, "
            f"range={start_date}:{end_date}."
        )

    out = raw[list(source_columns)].rename(columns=source_columns).copy()
    out["날짜"] = pd.to_datetime(out["날짜"], errors="coerce")
    for col in source_columns.values():
        if col != "날짜":
            out[col] = _krx_numeric(out[col])

    out = out.dropna(subset=["날짜", "기관합계", "외국인합계"])
    out = out.set_index("날짜").sort_index()
    if out.empty:
        raise KRXDataError(
            f"KRX investor flow has no valid rows for ticker={ticker}, "
            f"range={start_date}:{end_date}."
        )
    return out


def get_stock_features_today(ticker):
    ticker = str(ticker).strip().zfill(6)
    end_date = nearest_business_date_str()
    end = datetime.strptime(end_date, "%Y%m%d")
    start_date = (end - timedelta(days=500)).strftime("%Y%m%d")

    master = get_krx_stock_master()
    if ticker not in master.index:
        raise KRXDataError(
            f"ETF constituent ticker={ticker} is not present in the validated "
            f"KRX listed-stock master for {end_date}."
        )

    isin = master.loc[ticker, "isin"]
    name = master.loc[ticker, "name"]

    try:
        raw_px = retry_call(
            개별종목시세().fetch,
            start_date,
            end_date,
            isin,
            2,
            retries=CONFIG["api_max_retries"],
        )
        raw_flow = retry_call(
            투자자별_거래실적_개별종목_일별추이_일반().fetch,
            start_date,
            end_date,
            isin,
            2,
            3,
            retries=CONFIG["api_max_retries"],
        )
    except Exception as e:
        raise KRXDataError(
            f"KRX stock feature request failed for ticker={ticker}, "
            f"isin={isin}, range={start_date}:{end_date}: "
            f"{type(e).__name__}: {e}."
        ) from e

    px = parse_krx_stock_ohlcv(raw_px, ticker, start_date, end_date)
    flow = parse_krx_stock_flow(raw_flow, ticker, start_date, end_date)

    mcap = float(px["시가총액"].dropna().iloc[-1])
    if not np.isfinite(mcap) or mcap <= 0:
        raise KRXDataError(
            f"KRX market cap is invalid for ticker={ticker}, date={end_date}."
        )

    foreign_col = find_col(flow, "외국인")
    inst_col = find_col(flow, "기관")
    if foreign_col is None or inst_col is None:
        raise KRXDataError(
            f"KRX investor flow columns are missing for ticker={ticker}."
        )

    close = px["종가"].astype(float)
    high = px["고가"].astype(float)
    value = px["거래대금"].astype(float)
    foreign = flow[foreign_col].astype(float)
    institution = flow[inst_col].astype(float)

    feat = {
        "ticker": ticker,
        "name": name,
        "mcap": mcap,
    }

    for label, values in [
        ("foreign", foreign),
        ("institution", institution),
    ]:
        for n in [5, 20, 60]:
            raw = float(values.tail(n).sum()) if len(values) >= n else np.nan
            feat[f"{label}_{n}_to_mcap"] = (
                raw / mcap if pd.notna(raw) else np.nan
            )

    combined = foreign.add(institution, fill_value=0)
    roll20 = combined.rolling(20).sum() / mcap
    hist20 = roll20.dropna()
    current = hist20.iloc[-1] if len(hist20) else np.nan
    feat["flow_hist_percentile"] = (
        float((hist20 <= current).mean())
        if len(hist20) >= 20 and pd.notna(current)
        else np.nan
    )

    for n in [5, 20, 60, 120]:
        feat[f"ret{n}"] = return_n_days(close, n)

    ma20 = close.rolling(20).mean()
    ma60 = close.rolling(60).mean()
    ma120 = close.rolling(120).mean()
    feat["trend20_60"] = bool(
        len(close) >= 60 and close.iloc[-1] > ma20.iloc[-1] > ma60.iloc[-1]
    )
    feat["trend60_120"] = bool(
        len(close) >= 120 and ma60.iloc[-1] > ma120.iloc[-1]
    )

    feat["avg_value_20"] = value.tail(20).mean()
    avg5 = value.tail(5).mean()
    feat["value_expansion"] = (
        avg5 / feat["avg_value_20"]
        if feat["avg_value_20"] > 0
        else np.nan
    )

    high252 = high.tail(252).max()
    feat["dist_to_52w_high"] = (
        close.iloc[-1] / high252 - 1 if high252 > 0 else np.nan
    )
    dist_ma20 = (
        close.iloc[-1] / ma20.iloc[-1] - 1
        if pd.notna(ma20.iloc[-1])
        else np.nan
    )

    pullback = 0
    if feat["trend20_60"]:
        pullback += 30
        if pd.notna(dist_ma20):
            if -0.03 <= dist_ma20 <= 0.03:
                pullback += 30
            elif -0.06 <= dist_ma20 <= 0.06:
                pullback += 15
        if pd.notna(feat["ret5"]):
            if -0.08 <= feat["ret5"] <= 0:
                pullback += 25
            elif 0 < feat["ret5"] <= 0.03:
                pullback += 10
        if (
            pd.notna(feat["value_expansion"])
            and feat["value_expansion"] < 1
        ):
            pullback += 15

    feat["pullback_score"] = pullback
    row = px.iloc[-1]
    prev = px.iloc[-61:-1] if len(px) >= 61 else pd.DataFrame()

    if not prev.empty:
        open_price = float(row["시가"])
        day_high = float(row["고가"])
        day_close = float(row["종가"])
        prev_high = float(prev["고가"].max())
        body = abs(day_close - open_price)
        upper = day_high - max(open_price, day_close)
        feat["reversal_risk"] = bool(
            day_high >= prev_high
            and day_close < open_price
            and upper / max(body, 1e-9) >= 1.0
        )
    else:
        feat["reversal_risk"] = False

    return feat

def score_alpha(df):
    if df.empty:
        return df

    x = df.copy()

    cross = pd.Series(0.0,index=x.index)

    for investor, inv_w in CONFIG["investor_weights"].items():
        inv_score = pd.Series(0.0,index=x.index)

        for n, win_w in CONFIG["flow_window_weights"].items():
            col = f"{investor}_{n}_to_mcap"

            if col in x.columns:
                inv_score += (1-safe_pct_rank(x[col])).fillna(0)*100*win_w

        x[f"{investor}_empty_score"] = inv_score
        cross += inv_score*inv_w

    x["empty_cross_score"] = cross
    x["empty_hist_score"] = (1-x["flow_hist_percentile"])*100

    x["supply_empty_score"] = (
        0.65*x["empty_cross_score"] +
        0.35*x["empty_hist_score"].fillna(50)
    )

    x["price_momentum_score"] = (
        0.25*safe_pct_rank(x["ret20"]) +
        0.45*safe_pct_rank(x["ret60"]) +
        0.30*safe_pct_rank(x["ret120"])
    )*100

    x["trend_score"] = (
        60*x["trend20_60"].astype(int) +
        40*x["trend60_120"].astype(int)
    )

    x["liquidity_score"] = safe_pct_rank(x["value_expansion"])*100

    def high_score(v):
        if pd.isna(v): return 0
        if -0.05 <= v <= 0: return 100
        if -0.10 <= v < -0.05: return 80
        if -0.20 <= v < -0.10: return 60
        if -0.30 <= v < -0.20: return 30
        return 10

    x["high_proximity_score"] = x["dist_to_52w_high"].apply(high_score)

    x["heat_penalty"] = 0.0
    x.loc[x["ret20"]>0.12,"heat_penalty"] = 7
    x.loc[x["ret20"]>0.20,"heat_penalty"] = 15
    x.loc[x["ret20"]>0.30,"heat_penalty"] = 25

    w = CONFIG["alpha_weights"]

    x["alpha_score"] = (
        w["supply_empty"]*x["supply_empty_score"] +
        w["price_momentum"]*x["price_momentum_score"] +
        w["trend"]*x["trend_score"] +
        w["liquidity"]*x["liquidity_score"] +
        w["high_proximity"]*x["high_proximity_score"] +
        w["pullback"]*x["pullback_score"] -
        x["heat_penalty"] -
        x["reversal_risk"].astype(int)*35
    )

    x["eligible_alpha"] = (
        x["trend20_60"] &
        (x["avg_value_20"].fillna(0) >= CONFIG["min_stock_trading_value"])
    )

    return x.sort_values(
        ["eligible_alpha","alpha_score"],
        ascending=[False,False]
    ).reset_index(drop=True)


def build_alpha_candidates(leaders):
    all_rows = []
    stock_master = get_krx_stock_master()

    for _, r in leaders.iterrows():
        sector = r["sector"]
        etf_ticker = str(r["rep_ticker"])

        pdf = get_etf_constituents(etf_ticker)

        if pdf.empty:
            continue

        # ETF PDF can contain cash, futures, swaps, or foreign instruments whose
        # six-character PDF code happens to be numeric. Alpha is defined only
        # for currently listed domestic stocks validated by KRX ISU_CD.
        listed_mask = pdf["ticker"].isin(stock_master.index)
        non_stock = pdf.loc[~listed_mask].copy()
        if not non_stock.empty:
            name_col = (
                "구성종목명" if "구성종목명" in non_stock.columns else None
            )
            sample_parts = []
            for _, item in non_stock.head(5).iterrows():
                label = (
                    str(item[name_col]).strip()
                    if name_col is not None
                    else "unknown"
                )
                sample_parts.append(f"{item['ticker']}:{label}")
            print(
                "KRX_NON_STOCK_PDF_ITEMS_SKIPPED "
                f"source_etf={etf_ticker} sector={sector} "
                f"count={len(non_stock)} "
                f"sample={'|'.join(sample_parts)}"
            )

        pdf = pdf.loc[listed_mask].copy()
        if pdf.empty:
            print(
                "KRX_ETF_HAS_NO_LISTED_STOCK_CONSTITUENTS "
                f"source_etf={etf_ticker} sector={sector}"
            )
            continue

        if "비중" in pdf.columns:
            pdf = pdf.sort_values("비중", ascending=False)

        feats = []

        for t in pdf["ticker"].head(CONFIG["max_constituents_per_etf"]):
            # A validated listed ticker must fail explicitly if its market data
            # cannot be retrieved; it is not silently reclassified as non-stock.
            feat = get_stock_features_today(str(t))
            feats.append(feat)

        if not feats:
            continue

        scored = score_alpha(pd.DataFrame(feats))
        top = scored[scored["eligible_alpha"]].head(
            CONFIG["alpha_top_n_per_sector"]
        ).copy()

        if not top.empty:
            top["sector"] = sector
            top["source_etf"] = etf_ticker
            all_rows.append(top)

    return (
        pd.concat(all_rows, ignore_index=True)
        if all_rows
        else pd.DataFrame()
    )
````

# 9. Portfolio / Execution

````python

def load_holdings():
    path = BASE_DIR / "portfolio" / "current_holdings.csv"

    cols = [
        "ticker","name","sector","asset_type",
        "quantity","avg_price","current_price"
    ]

    if not path.exists():
        pd.DataFrame(columns=cols).to_csv(path,index=False,encoding="utf-8-sig")

    try:
        df = pd.read_csv(path, dtype={"ticker": "string"})
    except:
        return pd.DataFrame(columns=cols), path

    for c in cols:
        if c not in df.columns:
            df[c] = np.nan

    df["ticker"] = df["ticker"].astype("string").str.strip().str.upper()

    for c in ["quantity","avg_price","current_price"]:
        df[c] = pd.to_numeric(df[c],errors="coerce").fillna(0)

    df["market_value"] = df["quantity"]*df["current_price"]

    return df, path


def refresh_holdings_prices(holdings):
    if holdings.empty:
        return holdings

    out = holdings.copy()

    for idx, r in out.iterrows():
        p = get_latest_close_krx(str(r["ticker"]))
        if pd.notna(p) and p>0:
            out.loc[idx,"current_price"] = p

    out["market_value"] = out["quantity"]*out["current_price"]
    return out


def build_target_portfolio(leaders, alpha, stock_exposure):
    rows = []

    core_total = stock_exposure*CONFIG["etf_core_within_stock"]

    if not leaders.empty:
        etf_w = core_total/len(leaders)

        for _,r in leaders.iterrows():
            rows.append({
                "ticker":str(r["rep_ticker"]),
                "name":r["rep_name"],
                "sector":r["sector"],
                "asset_type":"ETF",
                "target_weight":etf_w,
                "alpha_score":np.nan,
            })

    if not alpha.empty:
        alpha_total = stock_exposure*CONFIG["alpha_within_stock"]
        eq_w = min(
            alpha_total/len(alpha),
            CONFIG["max_single_stock_weight"]
        )

        for _,r in alpha.iterrows():
            rows.append({
                "ticker":str(r["ticker"]),
                "name":r["name"],
                "sector":r["sector"],
                "asset_type":"STOCK",
                "target_weight":eq_w,
                "alpha_score":r["alpha_score"],
            })

    target = pd.DataFrame(rows)

    if target.empty:
        return target

    # sector cap
    sector_sum = target.groupby("sector")["target_weight"].sum()

    for sector,total in sector_sum.items():
        if total > CONFIG["max_sector_weight"]:
            mask = target["sector"]==sector
            target.loc[mask,"target_weight"] *= CONFIG["max_sector_weight"]/total

    return target


def build_rebalance_table(holdings,target,total_equity):
    current = holdings.copy()

    if current.empty:
        current = pd.DataFrame(columns=[
            "ticker","name","sector","asset_type",
            "quantity","current_price","market_value"
        ])

    current["current_weight"] = (
        current["market_value"]/total_equity
        if total_equity>0 else 0
    )

    merged = current.merge(
        target,
        on="ticker",
        how="outer",
        suffixes=("_current","_target")
    )

    for c in ["name","sector","asset_type"]:
        merged[c] = merged.get(f"{c}_target")
        if f"{c}_current" in merged.columns:
            merged[c] = merged[c].fillna(merged[f"{c}_current"])

    for c in ["quantity","market_value","current_weight","target_weight"]:
        merged[c] = pd.to_numeric(merged.get(c,0),errors="coerce").fillna(0)

    price_map = {}

    for _,r in current.iterrows():
        if r.get("current_price",0)>0:
            price_map[str(r["ticker"])] = float(r["current_price"])

    for t in merged["ticker"].astype(str):
        if t not in price_map:
            price_map[t] = get_latest_close_krx(t)

    merged["price"] = merged["ticker"].astype(str).map(price_map)
    merged["target_value"] = merged["target_weight"]*total_equity
    merged["trade_value"] = merged["target_value"]-merged["market_value"]
    merged["weight_gap"] = merged["target_weight"]-merged["current_weight"]

    def classify(r):
        if r["target_weight"]<=0 and r["current_weight"]>0:
            return "SELL"
        if r["current_weight"]<=0 and r["target_weight"]>0:
            return "BUY"
        if abs(r["weight_gap"])<=0.005:
            return "HOLD"
        return "BUY_MORE" if r["trade_value"]>0 else "REDUCE"

    merged["rebalance_action"] = merged.apply(classify,axis=1)

    return merged


ACTION_PRIORITY = {
    "SELL":1,
    "ROTATE":2,
    "REDUCE":3,
    "BUY":4,
    "BUY_MORE":5,
    "HOLD":99,
}


def approve_orders(rebalance,holdings,current_cash,total_equity,target_cash_weight):
    if rebalance.empty or total_equity<=0:
        return pd.DataFrame()

    x = rebalance.copy()
    x["priority"] = x["rebalance_action"].map(ACTION_PRIORITY).fillna(50)

    if "alpha_score" not in x.columns:
        x["alpha_score"] = np.nan

    x = x.sort_values(
        ["priority","alpha_score","trade_value"],
        ascending=[True,False,False]
    )

    turnover_budget = total_equity*CONFIG["max_daily_turnover"]
    used_turnover = 0.0

    available_cash = float(current_cash)
    min_cash_allowed = total_equity*max(
        target_cash_weight-CONFIG["cash_tolerance"],0
    )

    sector_weights = (
        holdings.groupby("sector")["market_value"].sum()/total_equity
    ).to_dict() if not holdings.empty else {}

    approved = []
    new_buy_count = 0

    for side_group in [["SELL","ROTATE","REDUCE"],["BUY","BUY_MORE"]]:
        for _,r in x.iterrows():
            action = r["rebalance_action"]

            if action not in side_group:
                continue

            price = r.get("price",np.nan)
            desired = abs(float(r.get("trade_value",0)))

            if not pd.notna(price) or price<=0 or desired<CONFIG["min_trade_value"]:
                continue

            remaining = turnover_budget-used_turnover

            if remaining<=0:
                break

            sector = r.get("sector",None)

            if action in ["SELL","ROTATE","REDUCE"]:
                qty = int(r.get("quantity",0))
                approve_value = min(desired,remaining)
                order_qty = min(qty,math.floor(approve_value/price))

                if order_qty<=0:
                    continue

                actual = order_qty*price
                signed_qty = -order_qty
                signed_value = -actual

                available_cash += actual

                if sector in sector_weights:
                    sector_weights[sector] = max(
                        0,
                        sector_weights[sector]-actual/total_equity
                    )

            else:
                if action=="BUY" and new_buy_count>=CONFIG["max_new_buys"]:
                    continue

                spendable = max(0,available_cash-min_cash_allowed)

                sector_room = max(
                    0,
                    (CONFIG["max_sector_weight"]-sector_weights.get(sector,0))*total_equity
                )

                approve_value = min(desired,remaining,spendable,sector_room)

                if approve_value<CONFIG["min_trade_value"]:
                    continue

                order_qty = math.floor(approve_value/price)

                if order_qty<=0:
                    continue

                actual = order_qty*price
                signed_qty = order_qty
                signed_value = actual

                available_cash -= actual
                sector_weights[sector] = sector_weights.get(sector,0)+actual/total_equity

                if action=="BUY":
                    new_buy_count += 1

            out = r.copy()
            out["approved_order_qty"] = signed_qty
            out["approved_trade_value"] = signed_value
            out["approved"] = True
            approved.append(out)

            used_turnover += abs(actual)

    if not approved:
        return pd.DataFrame()

    out = pd.DataFrame(approved)
    out["approved_turnover"] = used_turnover/total_equity
    out["post_cash_estimate"] = available_cash

    return out
````

# 10. NAV / Reporting

````python

def append_nav(holdings,current_cash,haa_state,target_cash_weight):
    path = BASE_DIR/"portfolio"/"nav_history.csv"
    today = datetime.now().strftime("%Y-%m-%d")

    stock_value = holdings["market_value"].sum() if not holdings.empty else 0

    etf_value = (
        holdings.loc[
            holdings["asset_type"].astype(str).str.upper()=="ETF",
            "market_value"
        ].sum()
        if not holdings.empty else 0
    )

    alpha_value = (
        holdings.loc[
            holdings["asset_type"].astype(str).str.upper()=="STOCK",
            "market_value"
        ].sum()
        if not holdings.empty else 0
    )

    row = pd.DataFrame([{
        "date":today,
        "total_equity":current_cash+stock_value,
        "cash":current_cash,
        "stock_value":stock_value,
        "etf_value":etf_value,
        "alpha_value":alpha_value,
        "haa_state":haa_state,
        "target_cash_weight":target_cash_weight,
    }])

    if path.exists():
        old = pd.read_csv(path)
        old = old[old["date"].astype(str)!=today]
        out = pd.concat([old,row],ignore_index=True)
    else:
        out = row

    out.sort_values("date").to_csv(path,index=False,encoding="utf-8-sig")
    return out.sort_values("date").reset_index(drop=True)


def nav_stats(nav):
    if len(nav)<2:
        return pd.Series(dtype=float),pd.DataFrame()

    x = nav.copy()
    x["date"] = pd.to_datetime(x["date"])
    x = x.sort_values("date").set_index("date")

    x["daily_return"] = x["total_equity"].pct_change()
    x["cum_return"] = x["total_equity"]/x["total_equity"].iloc[0]-1
    peak = x["total_equity"].cummax()
    x["drawdown"] = x["total_equity"]/peak-1

    r = x["daily_return"].dropna()
    vol = r.std(ddof=1)*np.sqrt(252) if len(r)>1 else np.nan
    sharpe = r.mean()*252/vol if pd.notna(vol) and vol>0 else np.nan

    days = max((x.index[-1]-x.index[0]).days,1)
    years = days/365.25
    cagr = (
        (x["total_equity"].iloc[-1]/x["total_equity"].iloc[0])**(1/years)-1
        if years>0 and x["total_equity"].iloc[0]>0
        else np.nan
    )

    s = pd.Series({
        "LatestNAV":x["total_equity"].iloc[-1],
        "TotalReturn":x["cum_return"].iloc[-1],
        "CAGR":cagr,
        "Volatility":vol,
        "Sharpe":sharpe,
        "MaxDrawdown":x["drawdown"].min(),
    })

    return s,x


def build_daily_report(result):
    rows = []

    rows += [
        {"section":"RISK","item":"HAA","value":result["haa"]["risk_state"].replace("_", "-")},
        {"section":"RISK","item":"TIP 13612U (Adj Close)","value":result["haa"]["tip_momentum"]},
        {"section":"RISK","item":"TIP 1M","value":result["haa"]["tip_1m"]},
        {"section":"RISK","item":"TIP 3M","value":result["haa"]["tip_3m"]},
        {"section":"RISK","item":"FearGreed","value":result["fear_greed"]["value"]},
        {"section":"ALLOCATION","item":"CashWeight","value":result["cash_info"]["cash_weight"]},
        {"section":"ALLOCATION","item":"StockExposure","value":1-result["cash_info"]["cash_weight"]},
        {"section":"SIGNAL","item":"LeaderCount","value":len(result["leaders"])},
        {"section":"SIGNAL","item":"AlphaCount","value":len(result["alpha_candidates"])},
        {"section":"EXECUTION","item":"ApprovedOrders","value":len(result["approved_orders"])},
    ]

    for k,v in result["nav_stats"].items():
        rows.append({"section":"PERFORMANCE","item":k,"value":v})

    return pd.DataFrame(rows)
````

# 12. 주문 없는 Full Daily Preview


````python
def run_daily_preview(current_cash=100_000_000):
    """
    빈 보유종목과 샘플 현금으로 전체 Daily 계산 경로를 검증한다.

    저장하지 않는 항목:
    - current_holdings.csv
    - ETF constituent snapshots
    - NAV history
    - reports / approved-order CSV
    - Google Drive files
    """
    current_cash = float(current_cash)
    if not np.isfinite(current_cash) or current_cash <= 0:
        raise ValueError("Daily Preview current_cash must be a positive number.")

    preflight = globals().get("AUTO_KRX_PREFLIGHT_RESULT")
    if preflight is None:
        preflight = krx_preflight(force=True)

    smoke = globals().get("MARKET_SMOKE_TEST_RESULT")
    if smoke is None:
        smoke = run_market_smoke_test()

    haa = smoke["_haa"]
    kospi = smoke["_kospi"]
    leaders = smoke["_leaders"].copy()
    kr_rank = smoke["_kr_rank"]
    us_rank = smoke["_us_rank"]

    fg = get_cnn_fear_greed()
    cash_info = tactical_cash_weight(
        haa["risk_on"],
        kospi,
        fg["value"],
    )
    stock_exposure = 1 - cash_info["cash_weight"]

    # 기존 Alpha 함수를 그대로 호출한다.
    alpha = build_alpha_candidates(leaders)
    if alpha.empty:
        raise RuntimeError(
            "Daily Preview produced no eligible Alpha candidates. "
            "No portfolio or order preview will be accepted."
        )

    holding_columns = [
        "ticker", "name", "sector", "asset_type",
        "quantity", "avg_price", "current_price", "market_value",
    ]
    holdings = pd.DataFrame(columns=holding_columns)
    total_equity = current_cash

    # 기존 ETF Core / Alpha 목표비중 함수를 그대로 호출한다.
    target = build_target_portfolio(
        leaders,
        alpha,
        stock_exposure,
    )
    if target.empty:
        raise RuntimeError("Daily Preview produced an empty target portfolio.")

    target["ticker"] = target["ticker"].astype(str).str.strip().str.upper()

    duplicate_tickers = sorted(
        target.loc[target["ticker"].duplicated(keep=False), "ticker"]
        .astype(str)
        .unique()
        .tolist()
    )
    if duplicate_tickers:
        raise RuntimeError(
            "Daily Preview found duplicate target tickers: "
            + ", ".join(duplicate_tickers)
        )

    if (target["target_weight"] < 0).any():
        raise RuntimeError("Daily Preview found a negative target weight.")

    if target["target_weight"].sum() > stock_exposure + 1e-9:
        raise RuntimeError(
            "Target weights exceed the strategy stock exposure."
        )

    stock_targets = target[
        target["asset_type"].astype(str).str.upper() == "STOCK"
    ]
    if (
        not stock_targets.empty
        and stock_targets["target_weight"].max()
        > CONFIG["max_single_stock_weight"] + 1e-9
    ):
        raise RuntimeError("Single-stock target weight cap was exceeded.")

    sector_weights = target.groupby("sector")["target_weight"].sum()
    if (
        not sector_weights.empty
        and sector_weights.max() > CONFIG["max_sector_weight"] + 1e-9
    ):
        raise RuntimeError("Target sector weight cap was exceeded.")

    rebalance = build_rebalance_table(
        holdings,
        target,
        total_equity,
    )
    if rebalance.empty:
        raise RuntimeError("Daily Preview produced an empty rebalance table.")

    actionable = rebalance["target_weight"] > 0
    invalid_price = (
        ~pd.to_numeric(
            rebalance.loc[actionable, "price"],
            errors="coerce",
        ).gt(0)
    )
    if invalid_price.any():
        bad = rebalance.loc[actionable].loc[invalid_price, "ticker"].astype(str)
        raise RuntimeError(
            "Daily Preview has missing/non-positive prices for: "
            + ", ".join(bad.tolist())
        )

    # 기존 주문 승인 함수를 그대로 호출하되 파일 저장은 하지 않는다.
    approved = approve_orders(
        rebalance,
        holdings,
        current_cash,
        total_equity,
        cash_info["cash_weight"],
    )
    if approved.empty:
        raise RuntimeError("Daily Preview produced no approved sample orders.")

    if (approved["approved_order_qty"] <= 0).any():
        raise RuntimeError(
            "Empty-holdings Preview must contain positive BUY quantities only."
        )

    new_buys = int((approved["rebalance_action"] == "BUY").sum())
    if new_buys > CONFIG["max_new_buys"]:
        raise RuntimeError("Daily Preview exceeded the new-buy limit.")

    approved_turnover = float(approved["approved_turnover"].iloc[-1])
    if approved_turnover > CONFIG["max_daily_turnover"] + 1e-9:
        raise RuntimeError("Daily Preview exceeded the turnover limit.")

    post_cash = float(approved["post_cash_estimate"].iloc[-1])
    min_cash_allowed = total_equity * max(
        cash_info["cash_weight"] - CONFIG["cash_tolerance"],
        0,
    )
    if post_cash + 1e-6 < min_cash_allowed:
        raise RuntimeError("Daily Preview violated the minimum cash reserve.")

    target_weight_sum = float(target["target_weight"].sum())
    sectors = ",".join(leaders["sector"].astype(str).tolist())

    print(
        "DAILY_PREVIEW_OK "
        f"sample_cash={current_cash:.0f} "
        f"risk_state={haa['risk_state']} "
        f"cash_weight={cash_info['cash_weight']:.4f} "
        f"leaders={len(leaders)} "
        f"alpha={len(alpha)} "
        f"target={len(target)} "
        f"orders={len(approved)} "
        f"new_buys={new_buys} "
        f"target_weight={target_weight_sum:.6f} "
        f"approved_turnover={approved_turnover:.6f} "
        f"post_cash={post_cash:.0f} "
        f"sectors={sectors}"
    )

    return {
        "status": "ok",
        "mode": "preview",
        "sample_cash": current_cash,
        "krx_preflight": preflight,
        "haa": haa,
        "fear_greed": fg,
        "cash_info": cash_info,
        "leaders": leaders,
        "kr_rank": kr_rank,
        "us_rank": us_rank,
        "alpha_candidates": alpha,
        "holdings": holdings,
        "target": target,
        "rebalance": rebalance,
        "approved_orders": approved,
        "approved_turnover": approved_turnover,
        "post_cash_estimate": post_cash,
        "persistent_writes": False,
        "run_daily_called": False,
    }


if os.environ.get("HAA_DAILY_PREVIEW", "0") == "1":
    preview_cash = float(
        os.environ.get("HAA_PREVIEW_CURRENT_CASH", "100000000")
    )
    DAILY_PREVIEW_RESULT = run_daily_preview(preview_cash)
````

# 13. RESEARCH ENTRYPOINT

````python

def rolling_13612u_series(close_series):
    s = close_series.dropna()
    out = pd.Series(index=s.index,dtype=float)

    for d in s.index:
        hist = s.loc[:d]
        vals = []

        for n in [21,63,126,252]:
            if len(hist)>n:
                vals.append(hist.iloc[-1]/hist.iloc[-(n+1)]-1)

        if vals:
            out.loc[d] = np.mean(vals)

    return out


def build_monthly_haa_regime(tip_close):
    mom = rolling_13612u_series(tip_close)
    rows = []

    for month_end in tip_close.resample("M").last().dropna().index:
        dates = tip_close.index[tip_close.index<=month_end]

        if len(dates)==0:
            continue

        signal_date = dates[-1]

        if signal_date not in mom.index or pd.isna(mom.loc[signal_date]):
            continue

        loc = tip_close.index.get_loc(signal_date)

        if loc+1>=len(tip_close.index):
            continue

        rows.append({
            "signal_date":signal_date,
            "apply_date":tip_close.index[loc+1],
            "risk_on":bool(mom.loc[signal_date]>0),
            "tip_13612u":mom.loc[signal_date],
        })

    return pd.DataFrame(rows)


def regime_log_to_daily(index,regime_log):
    s = pd.Series(False,index=index,dtype=bool)
    current = False

    apply_map = {
        r["apply_date"]:bool(r["risk_on"])
        for _,r in regime_log.iterrows()
        if r["apply_date"] in index
    }

    for d in index:
        if d in apply_map:
            current = apply_map[d]
        s.loc[d] = current

    return s


def performance_stats(result):
    if result is None or result.empty:
        return pd.Series(dtype=float)

    r = result["net_ret"].dropna()
    eq = result["equity"].dropna()

    if len(r)<2 or len(eq)<2:
        return pd.Series(dtype=float)

    years = max(len(r)/252,1/252)
    cagr = eq.iloc[-1]**(1/years)-1 if eq.iloc[-1]>0 else np.nan
    vol = r.std(ddof=1)*np.sqrt(252)
    sharpe = r.mean()*252/vol if vol>0 else np.nan
    dd = eq/eq.cummax()-1

    return pd.Series({
        "CAGR":cagr,
        "Volatility":vol,
        "Sharpe":sharpe,
        "MaxDrawdown":dd.min(),
        "TotalReturn":eq.iloc[-1]-1,
        "TurnoverPerDay":result.get("turnover",pd.Series(0,index=result.index)).mean(),
    })


def run_research():
    print("=== HAA Leader System v2.1 / RESEARCH ===")

    # 현재 연구 진입점은 v1.7까지 만든 구조를 연결하기 위한 준비 루틴
    # 장기 ETF 가격패널
    meta, kr_close = build_long_kr_etf_history()
    us_close = build_us_long_history()

    # HAA history
    haa_tickers = sorted(set(
        [CONFIG["haa_canary"]] +
        CONFIG["haa_offensive"] +
        CONFIG["haa_defensive"]
    ))

    haa_close = download_adjusted_close(haa_tickers, period="10y")
    tip = haa_close[CONFIG["haa_canary"]].dropna()
    haa_log = build_monthly_haa_regime(tip)
    haa_daily = regime_log_to_daily(kr_close.index,haa_log)

    result = {
        "kr_meta":meta,
        "kr_close":kr_close,
        "us_close":us_close,
        "haa_log":haa_log,
        "haa_daily":haa_daily,
    }

    print("KR ETF universe:",len(meta))
    print("KR history days:",len(kr_close))
    print("HAA regime rows:",len(haa_log))

    print(
        "\n다음 셀의 v1.6/v1.7 연구 함수들을 이용해 "
        "leader_history, Alpha snapshot, Walk-forward OOS를 실행할 수 있습니다."
    )

    return result
````


# 14. 연구용 함수 호환성

v2.1에서는 실전 코드를 깔끔하게 정리했지만,
v1.6~v1.7의 과거 주도섹터/Walk-forward 연구 함수는 그대로 재사용할 수 있습니다.

아래 셀은 이전 노트북의 연구 함수를 가져오는 대신,
필요한 핵심 함수만 재정의합니다.


````python

def make_walk_forward_windows(
    index,
    train_months=None,
    test_months=None,
):
    train_months = train_months or CONFIG["wf_train_months"]
    test_months = test_months or CONFIG["wf_test_months"]
    idx = pd.DatetimeIndex(index).sort_values().unique()

    if len(idx) == 0:
        return []

    start = idx.min().normalize()
    end = idx.max().normalize()

    windows = []
    train_start = start

    while True:
        train_end = train_start + pd.DateOffset(months=train_months) - pd.Timedelta(days=1)
        test_start = train_end + pd.Timedelta(days=1)
        test_end = test_start + pd.DateOffset(months=test_months) - pd.Timedelta(days=1)

        if test_start > end:
            break

        windows.append({
            "train_start": train_start,
            "train_end": min(train_end, end),
            "test_start": test_start,
            "test_end": min(test_end, end),
        })

        train_start = train_start + pd.DateOffset(months=test_months)

        if train_start >= end:
            break

    return windows
````

````python

HISTORICAL_ETF_PERIOD = "10y"

def build_long_kr_etf_history():
    universe = get_all_kr_etfs()
    universe = universe[
        (~universe["excluded"]) &
        (universe["sector"].notna())
    ].copy()

    yahoo_map = {f"{t}.KS": str(t) for t in universe["ticker"].astype(str)}
    close = download_close(yahoo_map.keys(), period=HISTORICAL_ETF_PERIOD)
    close = close.rename(columns=yahoo_map)

    available = set(close.columns.astype(str))
    meta = universe[universe["ticker"].astype(str).isin(available)].copy()

    return meta, close


def build_us_long_history():
    tickers = sorted({t for lst in US_SECTOR_ETFS.values() for t in lst})
    return download_close(tickers, period=HISTORICAL_ETF_PERIOD)
````

````python

def build_kr_etf_rank_asof(close_panel, meta_df, asof_date, min_history=70):
    asof = pd.Timestamp(asof_date)
    px = close_panel.loc[close_panel.index <= asof].copy()

    if px.empty:
        return pd.DataFrame()

    meta = meta_df.set_index("ticker")[["name","sector"]].to_dict("index")
    rows = []

    for ticker in px.columns:
        ticker = str(ticker)
        if ticker not in meta:
            continue

        s = px[ticker].dropna()
        if len(s) < min_history:
            continue

        last_date = pd.Timestamp(s.index[-1])
        if (asof - last_date).days > 10:
            continue

        r = s.pct_change().dropna()
        ret21 = return_n_days(s, 21)
        ret63 = return_n_days(s, 63)
        ret126 = return_n_days(s, 126)

        neg = r.tail(63)
        neg = neg[neg < 0]
        downside = neg.std(ddof=1) * np.sqrt(252) if len(neg) > 1 else np.nan
        sortino = ret63 / downside if pd.notna(downside) and downside > 0 else np.nan

        rows.append({
            "ticker": ticker,
            "name": meta[ticker]["name"],
            "sector": meta[ticker]["sector"],
            "ret21": ret21,
            "ret63": ret63,
            "ret126": ret126,
            "sortino63": sortino,
            "above_ma11": bool(s.iloc[-1] > s.rolling(11).mean().iloc[-1]),
            "above_ma21": bool(s.iloc[-1] > s.rolling(21).mean().iloc[-1]),
        })

    rank = pd.DataFrame(rows)
    if rank.empty:
        return rank

    rank["rs21_pct"] = rank["ret21"].rank(pct=True)
    rank["rs63_pct"] = rank["ret63"].rank(pct=True)
    rank["rs126_pct"] = rank["ret126"].rank(pct=True)
    rank["sortino_pct"] = rank["sortino63"].rank(pct=True)

    rank["momentum_score"] = (
        0.20 * rank["rs21_pct"] +
        0.50 * rank["rs63_pct"] +
        0.15 * rank["rs126_pct"] +
        0.15 * rank["sortino_pct"]
    )

    rank["trend_score"] = (
        0.6 * rank["above_ma11"].astype(int) +
        0.4 * rank["above_ma21"].astype(int)
    )

    rank["kr_score"] = 0.80 * rank["momentum_score"] + 0.20 * rank["trend_score"]
    return rank.sort_values("kr_score", ascending=False).reset_index(drop=True)
````

````python

def build_us_sector_rank_asof(close_panel, asof_date, min_history=70):
    asof = pd.Timestamp(asof_date)
    px = close_panel.loc[close_panel.index <= asof].copy()

    rows = []

    for sector, tickers in US_SECTOR_ETFS.items():
        tmp = []

        for t in tickers:
            if t not in px.columns:
                continue

            s = px[t].dropna()
            if len(s) < min_history:
                continue

            if (asof - pd.Timestamp(s.index[-1])).days > 10:
                continue

            r = s.pct_change().dropna()
            ret63 = return_n_days(s, 63)

            neg = r.tail(63)
            neg = neg[neg < 0]
            downside = neg.std(ddof=1) * np.sqrt(252) if len(neg) > 1 else np.nan
            sortino = ret63 / downside if pd.notna(downside) and downside > 0 else np.nan

            tmp.append({
                "ret63": ret63,
                "sortino": sortino,
                "above11": bool(s.iloc[-1] > s.rolling(11).mean().iloc[-1]),
                "above21": bool(s.iloc[-1] > s.rolling(21).mean().iloc[-1]),
            })

        if not tmp:
            continue

        x = pd.DataFrame(tmp)
        rows.append({
            "sector": sector,
            "us_ret63": x["ret63"].mean(),
            "us_sortino": x["sortino"].mean(),
            "us_above11": bool(x["above11"].mean() >= 0.5),
            "us_above21": bool(x["above21"].mean() >= 0.5),
        })

    rank = pd.DataFrame(rows)
    if rank.empty:
        return rank

    rank["us_rs63_pct"] = rank["us_ret63"].rank(pct=True)
    rank["us_sortino_pct"] = rank["us_sortino"].rank(pct=True)
    rank["us_score"] = (
        0.60 * rank["us_rs63_pct"] +
        0.20 * rank["us_sortino_pct"] +
        0.12 * rank["us_above11"].astype(int) +
        0.08 * rank["us_above21"].astype(int)
    )

    return rank.sort_values("us_score", ascending=False).reset_index(drop=True)
````

````python

def build_leaders_asof(
    kr_close_hist,
    kr_meta_hist,
    us_close_hist,
    asof_date,
    leader_count=None
):
    leader_count = leader_count or CONFIG["leader_sector_count"]
    kr_rank = build_kr_etf_rank_asof(
        kr_close_hist, kr_meta_hist, asof_date
    )
    us_rank = build_us_sector_rank_asof(
        us_close_hist, asof_date
    )

    if kr_rank.empty:
        return pd.DataFrame(), kr_rank, us_rank

    reps = (
        kr_rank.sort_values("kr_score", ascending=False)
        .groupby("sector", as_index=False)
        .first()
        [["sector","ticker","name","kr_score","ret63","above_ma11","above_ma21"]]
        .rename(columns={
            "ticker":"rep_ticker",
            "name":"rep_name",
            "ret63":"kr_ret63"
        })
    )

    avgs = (
        kr_rank.groupby("sector", as_index=False)
        .agg(
            kr_sector_avg_score=("kr_score","mean"),
            kr_etf_count=("ticker","count")
        )
    )

    out = reps.merge(avgs, on="sector", how="left")

    if not us_rank.empty:
        out = out.merge(
            us_rank[["sector","us_score","us_ret63","us_above11","us_above21"]],
            on="sector",
            how="left"
        )

    if "us_score" not in out.columns:
        out["us_score"] = 0.0
    if "us_above11" not in out.columns:
        out["us_above11"] = False

    out["us_score"] = out["us_score"].fillna(0)
    out["us_above11"] = out["us_above11"].fillna(False)

    out["cross_market_confirm"] = (
        out["us_above11"] &
        (out["us_score"] >= 0.60)
    )

    out["final_sector_score"] = (
        0.70 * out["kr_score"] +
        0.15 * out["kr_sector_avg_score"] +
        0.15 * out["cross_market_confirm"].astype(int)
    )

    out["eligible"] = out["above_ma11"]

    leaders = (
        out[out["eligible"]]
        .sort_values("final_sector_score", ascending=False)
        .head(leader_count)
        .reset_index(drop=True)
    )

    leaders["asof_date"] = pd.Timestamp(asof_date)
    return leaders, kr_rank, us_rank
````

````python

def build_leader_history(
    kr_close_hist,
    kr_meta_hist,
    us_close_hist,
    start=None,
    end=None,
    freq="W-FRI",
):
    idx = kr_close_hist.index

    if start is not None:
        idx = idx[idx >= pd.Timestamp(start)]
    if end is not None:
        idx = idx[idx <= pd.Timestamp(end)]

    if len(idx) == 0:
        return pd.DataFrame()

    weekly_dates = (
        pd.Series(index=idx, dtype=float)
        .resample(freq)
        .last()
        .index
    )

    rows = []

    for d in weekly_dates:
        hist = idx[idx <= d]
        if len(hist) == 0:
            continue

        signal_date = hist[-1]

        leaders, _, _ = build_leaders_asof(
            kr_close_hist,
            kr_meta_hist,
            us_close_hist,
            signal_date
        )

        for rank_no, (_, r) in enumerate(leaders.iterrows(), 1):
            rows.append({
                "signal_date": signal_date,
                "rank": rank_no,
                "sector": r["sector"],
                "rep_ticker": r["rep_ticker"],
                "rep_name": r["rep_name"],
                "final_sector_score": r["final_sector_score"],
                "cross_market_confirm": r["cross_market_confirm"],
            })

    return pd.DataFrame(rows)


def leader_turnover_report(leader_history):
    if leader_history.empty:
        return pd.DataFrame()

    grouped = (
        leader_history.groupby("signal_date")["sector"]
        .apply(list)
        .sort_index()
    )

    rows = []
    prev = None

    for d, sectors in grouped.items():
        cur = set(sectors)

        if prev is None:
            turnover = np.nan
            added, removed = [], []
        else:
            added = sorted(cur - prev)
            removed = sorted(prev - cur)
            turnover = len(cur.symmetric_difference(prev)) / max(len(cur | prev), 1)

        rows.append({
            "signal_date": d,
            "sectors": ",".join(sectors),
            "added": ",".join(added),
            "removed": ",".join(removed),
            "turnover": turnover,
        })

        prev = cur

    return pd.DataFrame(rows)
````

````python
def validate_actual_holdings_frame(raw):
    required = {
        "ticker", "name", "sector", "asset_type", "quantity", "avg_price",
    }
    if raw is None:
        raise ValueError("Actual holdings CSV input is missing.")

    missing = sorted(required - set(raw.columns))
    if missing:
        raise ValueError(
            "Actual holdings CSV is missing required columns: "
            + ", ".join(missing)
        )

    output_columns = [
        "ticker", "name", "sector", "asset_type",
        "quantity", "avg_price", "current_price", "market_value",
    ]
    if raw.empty:
        return pd.DataFrame(columns=output_columns)

    holdings = raw.copy()
    holdings["ticker"] = (
        holdings["ticker"].astype("string").str.strip().str.upper()
    )
    holdings["name"] = holdings["name"].astype("string").str.strip()
    holdings["sector"] = holdings["sector"].astype("string").str.strip()
    holdings["asset_type"] = (
        holdings["asset_type"].astype("string").str.strip().str.upper()
    )

    invalid_ticker = ~holdings["ticker"].str.fullmatch(
        r"[0-9A-Z]{6}", na=False
    )
    if invalid_ticker.any():
        bad = holdings.loc[invalid_ticker, "ticker"].astype(str).tolist()
        raise ValueError(
            "Actual holdings CSV has invalid tickers: " + ", ".join(bad)
        )

    duplicates = sorted(
        holdings.loc[
            holdings["ticker"].duplicated(keep=False), "ticker"
        ].astype(str).unique().tolist()
    )
    if duplicates:
        raise ValueError(
            "Actual holdings CSV has duplicate tickers: "
            + ", ".join(duplicates)
        )

    for col in ["quantity", "avg_price"]:
        holdings[col] = pd.to_numeric(holdings[col], errors="coerce")

    invalid_quantity = (
        holdings["quantity"].isna()
        | holdings["quantity"].lt(0)
        | holdings["quantity"].mod(1).ne(0)
    )
    if invalid_quantity.any():
        bad = holdings.loc[invalid_quantity, "ticker"].astype(str).tolist()
        raise ValueError(
            "Actual holdings CSV has invalid quantities for: "
            + ", ".join(bad)
        )

    holdings = holdings[holdings["quantity"].gt(0)].copy()
    if holdings.empty:
        return pd.DataFrame(columns=output_columns)

    invalid_avg = holdings["avg_price"].isna() | holdings["avg_price"].le(0)
    if invalid_avg.any():
        bad = holdings.loc[invalid_avg, "ticker"].astype(str).tolist()
        raise ValueError(
            "Actual holdings CSV has missing/non-positive avg_price for: "
            + ", ".join(bad)
        )

    invalid_text = (
        holdings["name"].isna()
        | holdings["name"].eq("")
        | holdings["sector"].isna()
        | holdings["sector"].eq("")
    )
    if invalid_text.any():
        bad = holdings.loc[invalid_text, "ticker"].astype(str).tolist()
        raise ValueError(
            "Actual holdings CSV has missing name/sector for: "
            + ", ".join(bad)
        )

    invalid_type = ~holdings["asset_type"].isin(["STOCK", "ETF"])
    if invalid_type.any():
        bad = holdings.loc[invalid_type, "ticker"].astype(str).tolist()
        raise ValueError(
            "Actual holdings asset_type must be STOCK or ETF for: "
            + ", ".join(bad)
        )

    holdings["quantity"] = holdings["quantity"].astype(int)
    holdings["current_price"] = 0.0
    holdings["market_value"] = 0.0
    return holdings[output_columns].reset_index(drop=True)


def load_actual_holdings_preview(path):
    source_path = Path(path).expanduser()
    if not source_path.is_file():
        raise FileNotFoundError(
            "Actual holdings CSV was not found: "
            f"{source_path}. Expected Google Drive file with columns "
            "ticker,name,sector,asset_type,quantity,avg_price."
        )

    try:
        raw = pd.read_csv(source_path, dtype={"ticker": "string"})
    except Exception as e:
        raise RuntimeError(
            f"Unable to read Actual holdings CSV {source_path}: "
            f"{type(e).__name__}: {e}"
        ) from e

    return validate_actual_holdings_frame(raw), source_path



ORDER_REVIEW_COLUMNS = [
    "ticker", "name", "sector", "portfolio_role", "rebalance_action",
    "reference_price", "order_qty", "estimated_order_value",
    "target_weight", "order_weight",
]


def build_order_review_table(approved, total_equity):
    if approved is None or approved.empty:
        return pd.DataFrame(columns=ORDER_REVIEW_COLUMNS)
    if not np.isfinite(total_equity) or total_equity <= 0:
        raise ValueError("Order Review total_equity must be positive.")

    required = {
        "ticker", "name", "sector", "asset_type", "rebalance_action",
        "price", "approved_order_qty", "approved_trade_value",
        "target_weight",
    }
    missing = sorted(required - set(approved.columns))
    if missing:
        raise ValueError(
            "Approved orders are missing Order Review columns: "
            + ", ".join(missing)
        )

    review = approved.copy()
    asset_type = review["asset_type"].astype(str).str.strip().str.upper()
    invalid_type = ~asset_type.isin(["STOCK", "ETF"])
    if invalid_type.any():
        bad = review.loc[invalid_type, "ticker"].astype(str).tolist()
        raise ValueError(
            "Order Review asset_type must be STOCK or ETF for: "
            + ", ".join(bad)
        )

    review["ticker"] = review["ticker"].astype(str).str.strip().str.upper()
    review["portfolio_role"] = np.where(
        asset_type.eq("ETF"), "ETF_CORE", "ALPHA"
    )
    review["reference_price"] = pd.to_numeric(
        review["price"], errors="coerce"
    )
    review["order_qty"] = pd.to_numeric(
        review["approved_order_qty"], errors="coerce"
    )
    review["estimated_order_value"] = pd.to_numeric(
        review["approved_trade_value"], errors="coerce"
    ).abs()
    review["target_weight"] = pd.to_numeric(
        review["target_weight"], errors="coerce"
    )
    review["order_weight"] = (
        review["estimated_order_value"] / float(total_equity)
    )

    invalid_numeric = (
        review["reference_price"].isna()
        | review["reference_price"].le(0)
        | review["order_qty"].isna()
        | review["order_qty"].eq(0)
        | review["estimated_order_value"].isna()
        | review["estimated_order_value"].le(0)
        | review["target_weight"].isna()
        | review["target_weight"].lt(0)
    )
    if invalid_numeric.any():
        bad = review.loc[invalid_numeric, "ticker"].astype(str).tolist()
        raise ValueError(
            "Order Review has invalid price/quantity/value/weight for: "
            + ", ".join(bad)
        )

    review["order_qty"] = review["order_qty"].astype(int)
    return review[ORDER_REVIEW_COLUMNS].reset_index(drop=True)

def run_actual_holdings_preview(current_cash, holdings_csv):
    """
    실제 현금·보유종목으로 Daily 계산 경로를 검증하되 어떤 파일도 쓰지 않는다.
    """
    current_cash = float(current_cash)
    if not np.isfinite(current_cash) or current_cash < 0:
        raise ValueError(
            "Actual Holdings Preview current_cash must be zero or positive."
        )

    preflight = globals().get("AUTO_KRX_PREFLIGHT_RESULT")
    if preflight is None:
        preflight = krx_preflight(force=True)

    smoke = globals().get("MARKET_SMOKE_TEST_RESULT")
    if smoke is None:
        smoke = run_market_smoke_test()

    holdings, source_path = load_actual_holdings_preview(holdings_csv)
    if holdings.empty:
        print("ACTUAL_HOLDINGS_INPUT_OK holdings=0 cash_only=True")
    holdings = refresh_holdings_prices(holdings)

    if (
        holdings["current_price"].isna().any()
        or holdings["current_price"].le(0).any()
    ):
        bad = holdings.loc[
            holdings["current_price"].isna()
            | holdings["current_price"].le(0),
            "ticker",
        ].astype(str).tolist()
        raise RuntimeError(
            "Actual Holdings Preview has invalid refreshed prices for: "
            + ", ".join(bad)
        )

    holdings_value = float(holdings["market_value"].sum())
    total_equity = current_cash + holdings_value
    if not np.isfinite(total_equity) or total_equity <= 0:
        raise ValueError(
            "Actual Holdings Preview total equity must be positive."
        )

    haa = smoke["_haa"]
    kospi = smoke["_kospi"]
    leaders = smoke["_leaders"].copy()
    kr_rank = smoke["_kr_rank"]
    us_rank = smoke["_us_rank"]

    fg = get_cnn_fear_greed()
    cash_info = tactical_cash_weight(
        haa["risk_on"],
        kospi,
        fg["value"],
    )
    stock_exposure = 1 - cash_info["cash_weight"]

    alpha = build_alpha_candidates(leaders)
    if alpha.empty:
        raise RuntimeError(
            "Actual Holdings Preview produced no eligible Alpha candidates."
        )

    target = build_target_portfolio(
        leaders,
        alpha,
        stock_exposure,
    )
    if target.empty:
        raise RuntimeError(
            "Actual Holdings Preview produced an empty target portfolio."
        )

    target["ticker"] = target["ticker"].astype(str).str.strip().str.upper()
    duplicates = sorted(
        target.loc[
            target["ticker"].duplicated(keep=False), "ticker"
        ].astype(str).unique().tolist()
    )
    if duplicates:
        raise RuntimeError(
            "Actual Holdings Preview found duplicate target tickers: "
            + ", ".join(duplicates)
        )

    if target["target_weight"].lt(0).any():
        raise RuntimeError(
            "Actual Holdings Preview found a negative target weight."
        )
    if target["target_weight"].sum() > stock_exposure + 1e-9:
        raise RuntimeError(
            "Actual target weights exceed the strategy stock exposure."
        )

    stock_targets = target[
        target["asset_type"].astype(str).str.upper().eq("STOCK")
    ]
    if (
        not stock_targets.empty
        and stock_targets["target_weight"].max()
        > CONFIG["max_single_stock_weight"] + 1e-9
    ):
        raise RuntimeError(
            "Actual Holdings Preview exceeded the single-stock cap."
        )

    sector_weights = target.groupby("sector")["target_weight"].sum()
    if (
        not sector_weights.empty
        and sector_weights.max()
        > CONFIG["max_sector_weight"] + 1e-9
    ):
        raise RuntimeError(
            "Actual Holdings Preview exceeded the sector cap."
        )

    rebalance = build_rebalance_table(
        holdings,
        target,
        total_equity,
    )
    if rebalance.empty:
        raise RuntimeError(
            "Actual Holdings Preview produced an empty rebalance table."
        )

    positioned = (
        rebalance["current_weight"].gt(0)
        | rebalance["target_weight"].gt(0)
    )
    invalid_price = ~pd.to_numeric(
        rebalance.loc[positioned, "price"],
        errors="coerce",
    ).gt(0)
    if invalid_price.any():
        bad = rebalance.loc[positioned].loc[
            invalid_price, "ticker"
        ].astype(str).tolist()
        raise RuntimeError(
            "Actual Holdings Preview has missing/non-positive prices for: "
            + ", ".join(bad)
        )

    approved = approve_orders(
        rebalance,
        holdings,
        current_cash,
        total_equity,
        cash_info["cash_weight"],
    )

    if approved.empty:
        approved_turnover = 0.0
        post_cash = current_cash
        new_buys = 0
    else:
        approved_turnover = float(
            approved["approved_turnover"].iloc[-1]
        )
        post_cash = float(approved["post_cash_estimate"].iloc[-1])
        new_buys = int(
            approved["rebalance_action"].eq("BUY").sum()
        )

        if approved_turnover > CONFIG["max_daily_turnover"] + 1e-9:
            raise RuntimeError(
                "Actual Holdings Preview exceeded the turnover limit."
            )
        if new_buys > CONFIG["max_new_buys"]:
            raise RuntimeError(
                "Actual Holdings Preview exceeded the new-buy limit."
            )

        buy_mask = approved["rebalance_action"].isin(["BUY", "BUY_MORE"])
        sell_mask = approved["rebalance_action"].isin(
            ["SELL", "ROTATE", "REDUCE"]
        )
        if (
            approved.loc[buy_mask, "approved_order_qty"].le(0).any()
            or approved.loc[sell_mask, "approved_order_qty"].ge(0).any()
        ):
            raise RuntimeError(
                "Actual Holdings Preview found an invalid order sign."
            )

        held_qty = holdings.set_index("ticker")["quantity"].to_dict()
        for _, order in approved.loc[sell_mask].iterrows():
            available = int(held_qty.get(str(order["ticker"]), 0))
            if abs(int(order["approved_order_qty"])) > available:
                raise RuntimeError(
                    "Actual Holdings Preview sell quantity exceeds holdings "
                    f"for ticker={order['ticker']}."
                )

    order_review = build_order_review_table(
        approved,
        total_equity,
    )
    if (
        not order_review.empty
        and abs(order_review["order_weight"].sum() - approved_turnover) > 1e-9
    ):
        raise RuntimeError(
            "Order Review weights do not match approved turnover."
        )

    min_cash_allowed = total_equity * max(
        cash_info["cash_weight"] - CONFIG["cash_tolerance"],
        0,
    )
    if post_cash < -1e-6:
        raise RuntimeError(
            "Actual Holdings Preview produced negative post-order cash."
        )
    if post_cash + 1e-6 < min_cash_allowed:
        raise RuntimeError(
            "Actual Holdings Preview violated the minimum cash reserve."
        )

    target_weight_sum = float(target["target_weight"].sum())
    sectors = ",".join(leaders["sector"].astype(str).tolist())

    print(
        "ACTUAL_HOLDINGS_PREVIEW_OK "
        f"cash={current_cash:.0f} "
        f"holdings={len(holdings)} "
        f"holdings_value={holdings_value:.0f} "
        f"total_equity={total_equity:.0f} "
        f"risk_state={haa['risk_state']} "
        f"cash_weight={cash_info['cash_weight']:.4f} "
        f"leaders={len(leaders)} "
        f"alpha={len(alpha)} "
        f"target={len(target)} "
        f"orders={len(approved)} "
        f"new_buys={new_buys} "
        f"target_weight={target_weight_sum:.6f} "
        f"approved_turnover={approved_turnover:.6f} "
        f"post_cash={post_cash:.0f} "
        f"sectors={sectors}"
    )

    if os.environ.get("HAA_ORDER_REVIEW_DETAIL", "0") == "1":
        printable = order_review.copy()
        if not printable.empty:
            printable["reference_price"] = printable[
                "reference_price"
            ].round().astype(int)
            printable["estimated_order_value"] = printable[
                "estimated_order_value"
            ].round().astype(int)
            printable["target_weight_pct"] = (
                printable["target_weight"] * 100
            ).round(2)
            printable["order_weight_pct"] = (
                printable["order_weight"] * 100
            ).round(2)
            printable = printable[
                [
                    "ticker", "name", "sector", "portfolio_role",
                    "rebalance_action", "reference_price", "order_qty",
                    "estimated_order_value", "target_weight_pct",
                    "order_weight_pct",
                ]
            ]

        print("ORDER_REVIEW_TABLE_BEGIN")
        if printable.empty:
            print("(no approved orders)")
        else:
            print(printable.to_string(index=False))
        print("ORDER_REVIEW_TABLE_END")
        print(
            "ORDER_REVIEW_PREVIEW_OK "
            f"orders={len(order_review)} "
            f"approved_turnover={approved_turnover:.6f} "
            f"persistent_writes=False run_daily_called=False"
        )

    return {
        "status": "ok",
        "mode": "actual_holdings_preview",
        "holdings_source": str(source_path),
        "current_cash": current_cash,
        "holdings_value": holdings_value,
        "total_equity": total_equity,
        "krx_preflight": preflight,
        "haa": haa,
        "fear_greed": fg,
        "cash_info": cash_info,
        "leaders": leaders,
        "kr_rank": kr_rank,
        "us_rank": us_rank,
        "alpha_candidates": alpha,
        "holdings": holdings,
        "target": target,
        "rebalance": rebalance,
        "approved_orders": approved,
        "order_review": order_review,
        "approved_turnover": approved_turnover,
        "post_cash_estimate": post_cash,
        "persistent_writes": False,
        "run_daily_called": False,
    }


if os.environ.get("HAA_ACTUAL_HOLDINGS_PREVIEW", "0") == "1":
    actual_cash_text = os.environ.get("HAA_CURRENT_CASH", "").strip()
    actual_holdings_path = os.environ.get(
        "HAA_HOLDINGS_CSV", ""
    ).strip()

    if not actual_cash_text:
        raise RuntimeError(
            "HAA_CURRENT_CASH is required for Actual Holdings Preview."
        )
    if not actual_holdings_path:
        raise RuntimeError(
            "HAA_HOLDINGS_CSV is required for Actual Holdings Preview."
        )

    ACTUAL_HOLDINGS_PREVIEW_RESULT = run_actual_holdings_preview(
        float(actual_cash_text),
        actual_holdings_path,
    )
````
